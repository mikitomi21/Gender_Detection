```python
import os
import shutil
from PIL import Image
from pathlib import Path
from PIL import Image
from autocrop import Cropper

current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)
data_dir = os.path.join(parent_dir, "data")
DATA_PATH = Path.cwd().parent / "data"

source_root = DATA_PATH / "processed"
target_root = DATA_PATH / "cropped"

categories = ['babies', 'men', 'none', 'women']
datasets = ['train', 'val', 'test']
```

```python
def process_image(img_path, save_path):
    cropper = Cropper()

    # Get a Numpy array of the cropped image
    cropped_array = cropper.crop(img_path)

    # Save the cropped image with PIL if a face was detected:
    if cropped_array is not None:
        centered_image = Image.fromarray(cropped_array)

        w, h = centered_image.size
        centered_image = centered_image.crop((100, 100, w-100, h-100))

        centered_image.save(save_path)
```

```python
for dataset in datasets:
    for category in categories:
        src_dir = os.path.join(source_root, dataset, category)
        dst_dir = os.path.join(target_root, dataset, category)

        # Make sure destination directory exists
        os.makedirs(dst_dir, exist_ok=True)

        for filename in os.listdir(src_dir):
            src_path = os.path.join(src_dir, filename)
            dst_path = os.path.join(dst_dir, filename)

            if os.path.isfile(src_path) and ".gitkeep" not in src_path:
                if category in ['men', 'women']:
                    process_image(src_path, dst_path)
                else:
                    shutil.copy2(src_path, dst_path)
```
